**But :** vérifier à l'œil que ChromaDB ramène des verbatims pertinents, et en tirer un premier chiffre de qualité (précision du retrieval).

**Avant :** avoir lancé `python -m rag.index` à la racine du projet (sinon la base est vide).

## 1. Se placer à la racine du projet et ouvrir la base

```python
import os
from pathlib import Path
# Remonte jusqu'au dossier qui contient le .env (marche sur Mac et sur Windows)
racine = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / ".env").exists())
os.chdir(racine)
from rag.embedder import embed
from rag.store import get_collection
col = get_collection("verbatims")
print(col.count(), "morceaux dans la base")
```

## 2. Trois questions en anglais

Plus la distance (entre crochets) est petite, plus le verbatim est proche de la question.
Pour chaque question, comptez à voix haute combien des 5 résultats sont **vraiment pertinents**.

```python
def search(question, k=5):
    res = col.query(query_embeddings=embed([question]), n_results=k)
    print("QUESTION :", question)
    for doc, dist in zip(res["documents"][0], res["distances"][0]):
        print(f"  [{dist:.2f}] {doc[:150]}")
    print()

search("nobody trained us on the tool")
search("afraid AI will replace my job")
search("management forces us to use AI")
```

## 3. Deux questions en français
Le modèle est multilingue : il doit trouver des verbatims anglais pertinents.

```python
search("on ne sait pas à quoi ça sert")
search("mon manager nous oblige à utiliser l'IA")
```
